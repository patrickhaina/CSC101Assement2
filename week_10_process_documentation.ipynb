## STEP 1: Understand the Problem

This program is designed to help restaurant staff quickly calculate tips for customers. The staff member enters the bill amount and the tip percentage, and the program calculates both the tip amount and the final total to charge the customer.

## STEP 2: Identify Inputs and Outputs

### INPUTS:
- Bill amount entered by the user
- Tip percentage entered by the user
- Both inputs are numerical values

### OUTPUTS:
- Tip amount in dollars
- Final total amount including the tip
- Results displayed in currency format

## STEP 3: Work It By Hand

Bill amount: $45.80  
Tip percentage: 15%

### Calculations:
1. Convert 15% into decimal:
   - 15% = 0.15

2. Multiply the bill amount by the tip percentage:
   - $45.80 × 0.15 = $6.87

3. Add the tip amount to the original bill:
   - $45.80 + $6.87 = $52.67

### Final Answer:
Tip amount is $6.87 and the total charge is $52.67.

# Problem 2: Late Fee Calculator

## STEP 1: Understand the Problem

This program helps library staff calculate overdue fines for borrowed books. The librarian enters how many days late the book is, and the program calculates the fine based on $0.50 per day, with a maximum fine limit of $10.00.

## STEP 2: Identify Inputs and Outputs

### INPUTS:
- Number of days the book is overdue
- Input is a numerical whole number

### OUTPUTS:
- Fine amount in dollars
- Message saying "Maximum fine reached" if the fine exceeds $10.00
- $0.00 fine if the book is returned on time

## STEP 3: Work It By Hand

### Case A: Book is 5 days late

Calculation:
- 5 × $0.50 = $2.50

Fine:
- $2.50

### Case B: Book is 25 days late

Calculation:
- 25 × $0.50 = $12.50

Since the maximum fine allowed is $10.00, the final fine becomes $10.00.

Fine:
- $10.00

Special Note:
- Maximum fine reached

### Case C: Book is 0 days late

Fine:
- $0.00
